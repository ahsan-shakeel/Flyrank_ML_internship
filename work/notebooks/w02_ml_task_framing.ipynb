## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Metric: Precision@K (Specifically Precision@50)

Why Precision@50?

Content editing teams operate under strict human review capacity constraints (e.g., reviewing 50 pages per week). Standard metrics like Accuracy or ROC-AUC can be misleading due to class imbalance (non-declining pages dominate). Precision@50 directly measures decision effectiveness: Of the top 50 pages recommended for review, how many actually required an update?

Secondary Metric: Average Precision (AP / PR-AUC) across the full ranking queue.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Static heuristics (e.g., "Flag pages older than 180 days" or "Flag pages with CTR < 1%") are rigid and imprecise:

High-authority evergreen pages may be 500 days old without needing updates.

Newer pages (60 days old) in fast-moving industries may decay rapidly.

Fixed CTR cutoffs fail to account for position tiering (Position 1 expectably gets a higher CTR than Position 9).

Why Machine Learning Wins:

ML models evaluate complex, non-linear interactions across multiple signals simultaneously—weighting search demand, expected CTR by position tier, engagement depth, and freshness decay—to produce a nuanced, data-driven ranking.
